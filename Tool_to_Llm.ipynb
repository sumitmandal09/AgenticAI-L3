```python
from langchain.tools import tool

@tool
def get_weather(city:str) -> str:
    """Get the weather for a city"""
    return f"The weather in {city} is sunny and warm."
    
```

```python
llm_with_tools = llm.bind_tools([get_weather])  #tool binding to the llm
```

```python
response =llm_with_tools.invoke("What is the weather in New York City?")
```

```python
response
```

```python
response.tool_calls
```

```python
for tool_call in response.tool_calls:
    print(f"Tool: {tool_call['name']}")
    print(f" Args: {tool_call['args']}")
    
```

```python
messages = [{"role": "user", "content": "What is the weather in New York City?"}]
ai_message = llm_with_tools.invoke(messages)
messages.append(ai_message) #
```

```python
for tool_call in ai_message.tool_calls:
    tool_message = get_weather.invoke(tool_call)
    
    messages.append(tool_message)
```

```python
response = llm_with_tools.invoke(messages)  
```
